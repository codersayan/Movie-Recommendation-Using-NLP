```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = {
    'title': [
        'Avengers', 'Iron Man', 'Thor', 'Captain America',
        'Interstellar', 'Inception', 'The Dark Knight',
        'The Matrix', 'Titanic', 'Avatar'
    ],
    
    'description': [
        'superhero team saves world action marvel',
        'genius builds iron suit superhero action marvel',
        'god of thunder superhero action marvel',
        'soldier becomes superhero patriotic action',
        'space time travel science fiction nasa',
        'dream within dream thriller mind bending',
        'batman fights joker crime dark city hero',
        'virtual reality simulation hacker action sci fi',
        'romantic tragedy ship sinking love story',
        'alien world exploration sci fi adventure'
    ]
}

df = pd.DataFrame(data)
df
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(df['description'])
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend(movie_name):
    if movie_name not in df['title'].values:
        return "Movie not found!"
    
    index = df[df['title'] == movie_name].index[0]
    
    similarity_scores = list(enumerate(similarity[index]))
    
    sorted_movies = sorted(similarity_scores, key=lambda x: x[1], reverse=True)
    
    print(f"\n🎬 Recommendations for {movie_name}:\n")
    
    count = 0
    for i in sorted_movies[1:6]:  # skip itself
        print(df.iloc[i[0]]['title'])
        count += 1
```

```python
recommend("Iron Man")
```

```python
data = {
    'title': [
        'Avengers', 'Iron Man', 'Thor', 'Captain America',
        'Interstellar', 'Inception', 'The Dark Knight',
        'The Matrix', 'Titanic', 'Avatar'
    ],

    'genre': [
        'action superhero', 'action superhero tech', 'action superhero myth',
        'action superhero war', 'sci-fi space', 'sci-fi thriller',
        'action crime thriller', 'sci-fi action AI', 'romance drama',
        'sci-fi adventure fantasy'
    ],

    'description': [
        'superhero team saves world action marvel',
        'genius builds iron suit superhero tech marvel',
        'god of thunder superhero action marvel mythology',
        'soldier becomes superhero patriotic war action',
        'space time travel black holes nasa exploration',
        'dream manipulation thriller mind bending reality',
        'batman fights joker crime dark city hero',
        'virtual reality simulation hacker machine world',
        'love story tragedy ship sinking emotional romance',
        'alien world exploration adventure sci fi planets'
    ]
}

df = pd.DataFrame(data)
df
```

```python
df['combined'] = df['genre'] + " " + df['description']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['combined'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend(movie_name, top_n=5):
    
    if movie_name not in df['title'].values:
        return "Movie not found!"

    idx = df[df['title'] == movie_name].index[0]
    scores = list(enumerate(similarity[idx]))

    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print(f"\n Recommendations for: {movie_name}\n")
    
    count = 0
    for i in scores[1:top_n+1]:
        movie_idx = i[0]
        score = round(i[1], 3)

        print(df.iloc[movie_idx]['title'])
        print("   Match Score:", score)

        # Explain WHY recommended
        print("   Why: similar genre/keywords →", df.iloc[movie_idx]['genre'])
        print("-"*50)

        count += 1
```

```python
recommend("Iron Man")
```

```python
df['popularity'] = [95, 90, 85, 88, 92, 94, 97, 89, 80, 91]
```

```python
df['popularity_score'] = df['popularity'] / 100
df.head()
```

```python
def hybrid_recommend(movie_name, top_n=5):
    
    if movie_name not in df['title'].values:
        return "Movie not found!"

    idx = df[df['title'] == movie_name].index[0]
    
    sim_scores = list(enumerate(similarity[idx]))

    results = []

    for i in sim_scores:
        movie_idx = i[0]
        
        content_score = i[1]
        popularity = df.iloc[movie_idx]['popularity_score']

        # weighted formula
        final_score = (0.7 * content_score) + (0.3 * popularity)

        results.append((movie_idx, final_score))

    results = sorted(results, key=lambda x: x[1], reverse=True)

    print(f"\n Hybrid Recommendations for: {movie_name}\n")

    for i in results[1:top_n+1]:
        idx = i[0]
        score = round(i[1], 3)

        print( df.iloc[idx]['title'])
        print("   Final Score:", score)
        print("   Genre:", df.iloc[idx]['genre'])
        print("-"*50)
```

```python
hybrid_recommend("Iron Man")
```

```python
liked_movies = ["Iron Man", "Inception"]
```

```python
def user_recommend(liked_movies, top_n=5):
    
    liked_indices = [df[df['title'] == m].index[0] for m in liked_movies]
    
    user_scores = similarity[liked_indices].mean(axis=0)

    results = list(enumerate(user_scores))
    results = sorted(results, key=lambda x: x[1], reverse=True)

    print("\n Personalized Recommendations:\n")

    for i in results[:top_n]:
        idx = i[0]

        print( df.iloc[idx]['title'])
        print("   Match:", round(i[1], 3))
        print("-"*40)
```

```python
user_recommend(liked_movies)
```

```python
def recommend_with_confidence(movie_name, top_n=5):
    
    idx = df[df['title'] == movie_name].index[0]
    scores = list(enumerate(similarity[idx]))
    
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print(f"\n Recommendations for {movie_name}\n")

    for i in scores[1:top_n+1]:
        idx = i[0]
        confidence = round(i[1] * 100, 2)

        print( df.iloc[idx]['title'])
        print(f"   Confidence: {confidence}%")
        print("-"*40)
```

```python
recommend_with_confidence("Thor")
```

```python
import pandas as pd

data = {
    'title': [
        'Inception', 'Interstellar', 'The Matrix',
        'Avengers', 'Iron Man', 'Titanic',
        'The Dark Knight', 'Avatar', 'Joker', 'Tenet'
    ],

    'overview': [
        'dreams within dreams mind bending thriller',
        'space time travel black holes nasa mission',
        'simulation reality hacker dystopian world',
        'superhero team saves world marvel action',
        'genius builds iron suit technology hero',
        'romantic tragedy ship sinks love story',
        'batman fights joker crime dark city',
        'alien world exploration sci fi adventure',
        'mental illness transformation dark character',
        'time inversion spy mission sci fi thriller'
    ],

    'genre': [
        'sci-fi thriller', 'sci-fi space', 'sci-fi action',
        'action superhero', 'action tech', 'romance drama',
        'action crime', 'sci-fi fantasy', 'drama crime', 'sci-fi action'
    ]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

df['combined'] = df['overview'] + " " + df['genre']

vectorizer = TfidfVectorizer(stop_words='english')
matrix = vectorizer.fit_transform(df['combined'])

similarity = cosine_similarity(matrix)
```

```python
def chatbot(query, top_n=5):
    query_vec = vectorizer.transform([query])
    scores = cosine_similarity(query_vec, matrix).flatten()

    top_indices = scores.argsort()[-top_n:][::-1]

    print("\n AI Movie Suggestions:\n")

    for i in top_indices:
        print( df.iloc[i]['title'])
        print("Genre:", df.iloc[i]['genre'])
        print("Match:", round(scores[i], 3))
        print("-"*40)
```

```python
chatbot("sci-fi space time travel")
```
